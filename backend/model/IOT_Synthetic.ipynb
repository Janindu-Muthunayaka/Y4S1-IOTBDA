```python
# =========================================
# PERFECT SYNTHETIC DATA GENERATOR
# Cold-Chain Fish Transport Dataset
# =========================================

import random
import numpy as np
import pandas as pd

random.seed(42)
np.random.seed(42)

data = []

def calculate_quality(avg_temp, max_temp, shock_count, max_accel, weight_loss, duration):
    score = 100

    # -----------------------------
    # Temperature Impact (MOST IMPORTANT)
    # -----------------------------
    if max_temp <= 2:
        score -= 0   # ideal
    elif max_temp <= 4:
        score -= 10
    elif max_temp <= 6:
        score -= 30
    else:
        score -= 50  # dangerous

    # -----------------------------
    # Shock / Handling Impact
    # -----------------------------
    if shock_count <= 5:
        score -= 0
    elif shock_count <= 12:
        score -= 10
    elif shock_count <= 20:
        score -= 25
    else:
        score -= 40

    # Acceleration penalty
    if max_accel > 4:
        score -= 15
    elif max_accel > 3:
        score -= 8

    # -----------------------------
    # Weight Loss Impact
    # -----------------------------
    if weight_loss <= 5:
        score -= 0
    elif weight_loss <= 12:
        score -= 10
    elif weight_loss <= 20:
        score -= 25
    else:
        score -= 40

    # -----------------------------
    # Duration Impact
    # -----------------------------
    if duration <= 120:
        score -= 0
    elif duration <= 200:
        score -= 10
    else:
        score -= 20

    # -----------------------------
    # Add small noise (important for ML realism)
    # -----------------------------
    score += random.uniform(-5, 5)

    return max(0, min(100, score))


# -----------------------------------------
# Generate Data
# -----------------------------------------
for _ in range(2000):   # increase if needed

    # Temperature generation (realistic patterns)
    base_temp = random.uniform(-2, 4)

    # Some trips go bad intentionally
    if random.random() < 0.25:
        base_temp = random.uniform(3, 8)

    avg_temp = base_temp
    max_temp = avg_temp + random.uniform(0, 3)
    min_temp = avg_temp - random.uniform(0, 2)

    # Shock data
    shock_count = random.randint(0, 30)
    max_accel = random.uniform(0.5, 5.0)

    # Weight loss
    weight_loss = random.uniform(0, 30)

    # Duration
    trip_duration = random.randint(30, 300)

    # Quality score
    quality_score = calculate_quality(
        avg_temp,
        max_temp,
        shock_count,
        max_accel,
        weight_loss,
        trip_duration
    )

    data.append([
        avg_temp,
        max_temp,
        min_temp,
        shock_count,
        max_accel,
        weight_loss,
        trip_duration,
        quality_score
    ])


# -----------------------------------------
# Create DataFrame
# -----------------------------------------
df = pd.DataFrame(data, columns=[
    "avg_temp",
    "max_temp",
    "min_temp",
    "shock_count",
    "max_accel",
    "weight_loss",
    "trip_duration",
    "quality_score"
])

# -----------------------------------------
# Add Classification Label (VERY IMPORTANT)
# -----------------------------------------
def classify(score):
    if score >= 80:
        return "SAFE"
    elif score >= 50:
        return "WARNING"
    else:
        return "CRITICAL"

df["risk_level"] = df["quality_score"].apply(classify)

# -----------------------------------------
# Save CSV
# -----------------------------------------
df.to_csv("perfect_cold_chain_dataset.csv", index=False)

print("✅ PERFECT dataset generated!")
print(df.head())

# -----------------------------------------
# Show class distribution
# -----------------------------------------
print("\nClass Distribution:")
print(df["risk_level"].value_counts())
```
